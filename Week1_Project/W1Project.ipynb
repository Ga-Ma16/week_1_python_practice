```python
import pandas as pd
import plotly.express as px

# --- 1. DATA ENGINE ---
df = pd.read_csv("most_subscribed_youtube_channels.csv")
df.columns = df.columns.str.strip()
df['subscribers'] = df['subscribers'].astype(str).str.replace(',', '')
df['video views'] = df['video views'].astype(str).str.replace(',', '')
df = df.dropna(subset=['subscribers', 'video views', 'category', 'started'])
df['subscribers'] = pd.to_numeric(df['subscribers'], errors='coerce')
df['video views'] = pd.to_numeric(df['video views'], errors='coerce')
df['started'] = df['started'].astype(int)

# --- 2. BUILD THE VISUALS ---

# Chart 1: Bar Chart
top_10 = df.nlargest(10, 'subscribers')
fig1 = px.bar(top_10, x='subscribers', y='Youtuber', orientation='h', color='category', 
              title="🏆 Top 10 Global Channels by Subscribers", template="plotly_dark")
fig1.update_layout(yaxis={'categoryorder':'total ascending'})

# Chart 2: Pie Chart (Donut style for a sleeker UI)
fig2 = px.pie(df, names='category', title="📊 Channel Distribution by Category", 
              hole=0.4, template="plotly_dark")

# Chart 3: Box Plot (Views spread across categories)
fig3 = px.box(df, x='category', y='video views', color='category', 
              title="📈 Video Views Spread Across Categories", template="plotly_dark")
fig3.update_layout(xaxis={'categoryorder':'median descending'}) 

# --- 3. COMPILE THE DASHBOARD ---
# We stitch all three charts into a single, sleek HTML page
html_dashboard = f"""
<html>
<head><title>YouTube Analytics</title></head>
<body style="background-color: #0F0F0F; color: white; font-family: sans-serif; text-align: center;">
    <h1 style="padding: 20px; color: #66FCF1; text-shadow: 0 0 10px rgba(69, 162, 158, 0.7);">
        🔮 YouTube Analytics Dashboard
    </h1>
    <div style="width: 85%; margin: auto;">
        {fig1.to_html(full_html=False, include_plotlyjs='cdn')}
        <br><br>
        {fig2.to_html(full_html=False, include_plotlyjs=False)}
        <br><br>
        {fig3.to_html(full_html=False, include_plotlyjs=False)}
    </div>
</body>
</html>
"""

# Save the final file
with open("dashboard.html", "w", encoding="utf-8") as f:
    f.write(html_dashboard)

print("✅ Mega-Dashboard Compiled! Double click 'dashboard.html' on the left to view.")
```
